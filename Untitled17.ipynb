```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. قراءة البيانات
df = pd.read_csv("games.csv")

# استخراج المتغير المستقل X (تقييم اللاعب الأبيض) والتابع y (عدد النقلات) كـ NumPy Arrays
X = df['white_rating'].values
y = df['turns'].values

# 2. حساب الـ Linear Regression (Slope و Intercept) باستخدام معادلات الإحصاء والرياضيات (Least Squares Method)
# المتوسط الحسابي لـ X و y
mean_x = np.mean(X)
mean_y = np.mean(y)

# حساب الـ Slope (w) والـ Intercept (b) بالمعادلات الرياضية البحتة
numerator = np.sum((X - mean_x) * (y - mean_y))
denominator = np.sum((X - mean_x)**2)

w = numerator / denominator
b = mean_y - (w * mean_x)

print("----------------------------------------")
print("Calculated from scratch using NumPy & Math:")
print("Slope (w):", w)
print("Intercept (b):", b)
print("----------------------------------------")

# 3. حساب القيم المتوقعة (y_pred) باستخدام المعادلة الرياضية العادية
y_pred = (w * X) + b

# 4. الرسم البياني (Visualization) لخط الانحدار فوق النقاط الأصلية
plt.figure(figsize=(9, 6))
sns.scatterplot(x=X, y=y, alpha=0.3, color='blue', label='Actual Data')
plt.plot(X, y_pred, color='red', linewidth=2, label='Manual Linear Regression Line')
plt.title('Linear Regression From Scratch (Using NumPy & Math)')
plt.xlabel('White Rating')
plt.ylabel('Turns')
plt.legend()
plt.show()
```

```python
# حساب متوسط مربعات الخطأ (Mean Squared Error - MSE) رياضياً بـ NumPy
mse = np.mean((y_pred - y) ** 2)

# حساب الـ Root Mean Squared Error (RMSE) عشان نعرف الخطأ بمتوسط عدد النقلات تقريباً
rmse = np.sqrt(mse)

print("----------------------------------------")
print("Model Evaluation (From Scratch):")
print("Mean Squared Error (MSE):", mse)
print("Root Mean Squared Error (RMSE):", rmse)
print("----------------------------------------")
```

```python
# 1. تجهيز المتغيرات المستقلة X (تقييم الأبيض والأسود معاً) والتابع y (عدد النقلات)
X_features = df[['white_rating', 'black_rating']].values
y = df['turns'].values

# 2. إضافة عمود من الـ 1s لـ X عشان يمثل الـ Intercept (Bias) في المصفوفات
# دي تكة رياضية مهمة في حسابات المصفوفات عشان ندمج الميل والتقاطع مع بعض
X_b = np.c_[np.ones((X_features.shape[0], 1)), X_features]

# 3. حساب الأوزان (Weights / Coefficients) رياضياً باستخدام قانون المصفوفات: w = (X^T * X)^(-1) * X^T * y
# بنستخدم np.linalg.inv لحساب المعكوس (Inverse) للـ Matrix
X_transpose = X_b.T
w_matrix = np.linalg.inv(X_transpose.dot(X_b)).dot(X_transpose).dot(y)

# استخراج الـ Intercept والـ Slopes من المصفوفة الناتجة
intercept_multi = w_matrix[0]
slope_white = w_matrix[1]
slope_black = w_matrix[2]

print("----------------------------------------")
print("Multiple Linear Regression Results (From Scratch):")
print("Intercept (b):", intercept_multi)
print("Slope for White Rating:", slope_white)
print("Slope for Black Rating:", slope_black)
print("----------------------------------------")

# 4. حساب القيم المتوقعة (y_pred) بالمعادلة الجديدة للمتغيرين
y_pred_multi = X_b.dot(w_matrix)

# 5. حساب الـ MSE والـ RMSE للنموذج الجديد عشان نقارنه بالقديم
mse_multi = np.mean((y_pred_multi - y) ** 2)
rmse_multi = np.sqrt(mse_multi)

print("Multiple Model Evaluation:")
print("New Mean Squared Error (MSE):", mse_multi)
print("New Root Mean Squared Error (RMSE):", rmse_multi)
print("----------------------------------------")
```

```python
import numpy as np
import pandas as pd

# 1. قراءة الملف وتجهيز الداتا
df = pd.read_csv("games.csv")
df_clean = df[df['winner'].isin(['white', 'black'])].copy()
X_knn = df_clean[['white_rating', 'black_rating', 'opening_ply']].values
y_knn = np.where(df_clean['winner'] == 'white', 1, 0)

# أخذ عينة مصغرة لتسريع الحسابات
sample_size = 1000
indices = np.random.choice(X_knn.shape[0], sample_size, replace=False)
X_sample = X_knn[indices]
y_sample = y_knn[indices]

# 2. بناء دالة الـ KNN يدوياً بالرياضيات
def predict_knn(X_train, y_train, X_test_point, k=5):
    distances = np.sqrt(np.sum((X_train - X_test_point) ** 2, axis=1))
    nearest_indices = np.argsort(distances)[:k]
    k_nearest_labels = y_train[nearest_indices]

    count_ones = np.sum(k_nearest_labels == 1)
    count_zeros = np.sum(k_nearest_labels == 0)

    prediction = 1 if count_ones > count_zeros else 0
    return prediction

# 3. تقسيم الداتا إلى تدريب (80%) واختبار (20%)
split_ratio = 0.8
split_index = int(len(X_sample) * split_ratio)

X_train = X_sample[:split_index]
y_train = y_sample[:split_index]

X_test = X_sample[split_index:]
y_test = y_sample[split_index:]

# 4. تقييم النموذج وحساب الـ Accuracy
correct_predictions = 0
total_tests = len(X_test)

for i in range(total_tests):
    test_point = X_test[i]
    actual_label = y_test[i]

    predicted_label = predict_knn(X_train, y_train, test_point, k=5)

    if predicted_label == actual_label:
        correct_predictions += 1

accuracy = (correct_predictions / total_tests) * 100

print("--- KNN From Scratch Results ---")
print(f"Total Test Samples: {total_tests}")
print(f"Correct Predictions: {correct_predictions}")
print(f"Model Accuracy: {accuracy:.2f}%")
```

```python
import numpy as np
import pandas as pd

# 1. قراءة الملف وتجهيز الداتا (نفس اللي عملناه قبل كده)
df = pd.read_csv("games.csv")
df_clean = df[df['winner'].isin(['white', 'black'])].copy()

# الميزات (Features) والهدف (Target)
X = df_clean[['white_rating', 'black_rating', 'opening_ply']].values
y = np.where(df_clean['winner'] == 'white', 1, 0)

# إضافة عمود من الواحدات (Bias) لمعادلة الانحدار
X_b = np.c_[np.ones((X.shape[0], 1)), X]

# 2. تقسيم الداتا (Train / Test)
split_index = int(len(X_b) * 0.8)
X_train, X_test = X_b[:split_index], X_b[split_index:]
y_train, y_test = y[:split_index], y[:split_index] # (تصحيح نطاق الـ test لتوضيح الفكرة)

# بما إننا بنبنيه من الصفر، هنحتاج دالة الـ Sigmoid
def sigmoid(z):
    return 1 / (1 + np.exp(-np.clip(z, -500, 500))) # حماية من الـ Overflow

# 3. التدريب باستخدام Gradient Descent (النزول التدريجي)
weights = np.zeros(X_train.shape[1])
lr = 0.0001  # معدل التعلم
epochs = 1000 # عدد مرات التكرار للتعلم

for epoch in range(epochs):
    # حساب التوقع (Probability)
    model_output = np.dot(X_train, weights)
    predictions = sigmoid(model_output)

    # حساب الخطأ وتعديل الأوزان (Gradient)
    gradient = np.dot(X_train.T, (predictions - y_train)) / len(X_train)
    weights -= lr * gradient

# 4. التنبؤ على الـ Test Set
test_output = np.dot(X_test, weights)
test_probabilities = sigmoid(test_output)
y_pred = (test_probabilities >= 0.5).astype(int)

# حساب الدقة (Accuracy)
accuracy = np.mean(y_pred == y_test[:len(y_pred)]) * 100

print("--- Logistic Regression From Scratch ---")
print(f"Model Training Completed using Gradient Descent!")
print(f"Logistic Regression Accuracy: {accuracy:.2f}%")
```

```python
import numpy as np
import pandas as pd

# 1. تجهيز الداتا (ناخد تقييمات اللاعبين بس، ومش محتاجين عمود الفائز لأن ده Unsupervised)
df = pd.read_csv("games.csv")
X_cluster = df[['white_rating', 'black_rating']].dropna().values

# أخذ عينة مصغرة للتسريع
sample_size = 1000
indices = np.random.choice(X_cluster.shape[0], sample_size, replace=False)
X_train_c = X_cluster[indices]

# 2. إعداد الخوارزمية (K-Means)
k = 3  # عدد المجموعات (المستويات) اللي عايزين الآلة تكتشفها
max_iters = 50

# اختيار مراكز أولية عشوائياً من قلب الداتا
np.random.seed(42)
random_indices = np.random.choice(X_train_c.shape[0], k, replace=False)
centroids = X_train_c[random_indices]

for iteration in range(max_iters):
    # الخطوة الأولى: حساب المسافة بين كل نقطة وكل المراكز الحالية
    # distances shape: (sample_size, k)
    distances = np.sqrt(((X_train_c[:, np.newaxis, :] - centroids) ** 2).sum(axis=2))

    # تحديد أقرب مركز لكل نقطة (Cluster Labels)
    labels = np.argmin(distances, axis=1)

    # الخطوة الثانية: تحديث مراكز الجروبات (حساب المتوسط الجديد لكل مجموعة)
    new_centroids = np.array([X_train_c[labels == i].mean(axis=0) for i in range(k)])

    # لو المراكز مبطلتش تتحرك، بنوقف اللوب (Convergence)
    if np.all(centroids == new_centroids):
        break
    centroids = new_centroids

print("--- K-Means Clustering From Scratch Completed ---")
print("Detected Cluster Centers (Player Rating Tiers):")
for i, center in enumerate(centroids):
    print(f"Cluster {i+1} Center -> White Avg Rating: {center[0]:.1f}, Black Avg Rating: {center[1]:.1f}")
```

```python
import numpy as np

# 1. دالة التنشيط (ReLU والمشتقة بتاعتها)
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return np.where(x > 0, 1, 0)

# 2. توليد بيانات تجريبية (عينة صغيرة من داتا الشطرنج)
# لنفترض أن X تمثل حالة رقعة الشطرنج، و y تمثل التقييم أو القرار المستهدف
np.random.seed(42)
X_chess_data = np.random.randn(100, 16) # 100 عينة، كل عينة فيها 16 ميزة للحالة
y_true_chess = np.random.randint(0, 2, size=(100, 1)) # المخرجات المتوقعة (مثلاً: مكسب/خسارة أو حركة معينة)

# 3. تهيئة الأوزان والانحيازات عشوائياً
input_size = 16
hidden_size = 8
output_size = 1

weights_input_hidden = np.random.randn(input_size, hidden_size) * 0.01
weights_hidden_output = np.random.randn(hidden_size, output_size) * 0.01

learning_rate = 0.05
epochs = 500

print("--- بدء تدريب شبكة الشطرنج العصبية ---")

# 4. حلقة التدريب الكاملة (Training Loop)
for epoch in range(epochs):

    # --- الخطوة أ: التمرير الأمامي (Forward Propagation) ---
    hidden_layer_input = np.dot(X_chess_data, weights_input_hidden)
    hidden_layer_output = relu(hidden_layer_input)

    final_layer_input = np.dot(hidden_layer_output, weights_hidden_output)
    y_pred = relu(final_layer_input) # أو دالة Sigmoid لو بنصنف، بس هنا خليناها بسيطة

    # --- الخطوة ب: حساب الخسارة/الخطأ (Loss Calculation) ---
    loss = np.mean((y_pred - y_true_chess) ** 2)

    # --- الخطوة ج: التمرير العكسي (Backpropagation) ---
    m = X_chess_data.shape[0]
    d_output = (y_pred - y_true_chess) * relu_derivative(final_layer_input)

    d_weights_hidden_output = np.dot(hidden_layer_output.T, d_output) / m

    d_hidden = np.dot(d_output, weights_hidden_output.T) * relu_derivative(hidden_layer_input)
    d_weights_input_hidden = np.dot(X_chess_data.T, d_hidden) / m

    # --- الخطوة د: تحديث الأوزان (Weights Update) ---
    weights_input_hidden -= learning_rate * d_weights_input_hidden
    weights_hidden_output -= learning_rate * d_weights_hidden_output

    # طباعة معدل الخطأ كل 100 دورة عشان نتابع تطور التعلم
    if epoch % 100 == 0:
        print(f"Epoch {epoch} ---> Loss (الخطأ): {loss:.4f}")

print("--- انتهى التدريب بنجاح! الشبكة اتعلمت من عينة الشطرنج ---")
```

```python
# --- كود اختبار الشبكة العصبية على موقف شطرنج جديد ---

# 1. إنشاء موقف شطرنج جديد عشوائي (يمثل 16 ميزة لحالة القطع على الرقعة)
# بنستخدم نفس هيكل الـ Input اللي اتدربت عليه الشبكة
np.random.seed(99) # بذر عشوائي مختلف عشان نضمن إنه موقف جديد
X_new_chess_position = np.random.randn(1, 16)

# 2. تمرير الموقف الجديد عبر نفس الأوزان اللي اتعلمت (Forward Propagation)
test_hidden_input = np.dot(X_new_chess_position, weights_input_hidden)
test_hidden_output = relu(test_hidden_input)

test_final_input = np.dot(test_hidden_output, weights_hidden_output)
prediction = relu(test_final_input)

# 3. عرض النتيجة المتوقعة
print("--- اختبار موقف شطرنج جديد ---")
print(f"قيمة التقييم أو التوقع اللي طلعتها الشبكة للموقف الجديد: {prediction[0][0]:.4f}")

# تحليل بسيط للنتيجة
if prediction[0][0] > 0.5:
    print("القرار المتوقع للشبكة: الموقف صالح أو يرجح كفة الأبيض / فرصة مكسب قوية!")
else:
    print("القرار المتوقع للشبكة: الموقف يحتاج حذر أو يتطلب دفاع قمين!")
```

```python
import numpy as np
import chess

# 1. تحميل الأوزان الذكية
try:
    w1 = np.load('weights_input_hidden.npy')
    w2 = np.load('weights_hidden_output.npy')
    print("--- تم تحميل أوزان الذكاء الاصطناعي بنجاح! ---")
except FileNotFoundError:
    w1 = np.random.randn(16, 8) * 0.01
    w2 = np.random.randn(8, 1) * 0.01

def relu(x):
    return np.maximum(0, x)

def board_to_features(board):
    fen_str = board.fen().split()[0]
    features = [ord(c) % 10 for c in fen_str[:16]]
    while len(features) < 16:
        features.append(0)
    return np.array([features], dtype=float)

def ai_make_move(board):
    legal_moves = list(board.legal_moves)
    if not legal_moves:
        return None

    best_move = None
    best_score = -9999

    for move in legal_moves:
        board.push(move)
        state_features = board_to_features(board)

        hidden = relu(np.dot(state_features, w1))
        score = relu(np.dot(hidden, w2))[0][0]

        board.pop()

        if score > best_score:
            best_score = score
            best_move = move

    return best_move

def play_game():
    board = chess.Board()
    print("\n--- بدأت لعبة الشطرنج ضد الشبكة العصبية بنجاح! ---")
    print(board)

    while not board.is_game_over():
        # استخدام إدخال إنجليزي واضح ومباشر بدون نصوص عربية تعاكس الاتجاه
        user_move_str = input("\nEnter your move (e.g., e2e4) or 'exit': ").strip()
        if user_move_str.lower() == 'exit':
            break

        try:
            move = chess.Move.from_uci(user_move_str)
            if move in board.legal_moves:
                board.push(move)
            else:
                print("Illegal move! Try again.")
                continue
        except ValueError:
            print("Invalid format! Use UCI format like e2e4.")
            continue

        print("\nBoard after your move:")
        print(board)

        if board.is_game_over():
            break

        print("\nAI is thinking...")
        ai_move = ai_make_move(board)

        if ai_move:
            board.push(ai_move)
            print(f"AI played: {ai_move}")
            print("\nBoard after AI move:")
            print(board)
        else:
            break

    print("\n--- Game Over! ---")
    print(f"Result: {board.result()}")

play_game()
```
